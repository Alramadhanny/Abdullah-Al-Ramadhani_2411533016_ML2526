```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.metrics import mean_squared_error
```

```python
DATA_PATH = "/content/drive/MyDrive/Dataset/advertising (1).csv"
df = pd.read_csv(DATA_PATH)

x = df.drop("Sales", axis=1)
y = df["Sales"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)

rmse_train = np.sqrt(mean_squared_error(y_train, y_pred_train))
rmse_test = np.sqrt(mean_squared_error(y_test, y_pred_test))

print("=== Holdout Method ===")
print("RMSE on Training Set:", rmse_train)
print("RMSE on Testing Set:", rmse_test)

```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)

rmse_list= []

for train_index, test_index in kf.split(x):
  X_train, X_test = x.iloc[train_index], x.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  model_k = LinearRegression()
  model_k.fit(X_train, y_train)

  y_pred = model_k.predict(X_test)
  rmse = np.sqrt(mean_squared_error(y_test, y_pred))

  rmse_list.append(rmse)

print("\n=== K-Fold Cross Validation ===")
print("RMSE tiap fold:", rmse_list)
print("RMSE rata-rata:", np.mean(rmse_list))
```

```python
cv_scores = np.sqrt(
    -cross_val_score(
        LinearRegression(),
        x, y,
        cv=5,
        scoring="neg_mean_squared_error")
    )

print("\n=== K-Fold dengan cross_val_score ===")
print("RMSE tiap fold:", cv_scores)
print("RMSE rata-rata:", np.mean(cv_scores))
```
